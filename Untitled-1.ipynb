```python
#Initialize a list of the first 100 integers starting from 0 up until 99
random_list = np.arange(100)
```

```python
#Access elements within the array
print(random_list[0])
```

```python
# the: operator which can be used to access elements across ranges
print(random_list[5:15])
```

```python
# use: at the start of end of an array to print out "everything from start until" and "everything from ___ until end"
print(random_list[:5])
```

```python
print(random_list[5:])
```

```python
# We can also do math on our aray
random_list = random_list + 1
print(random_list)
random_list = random_list * 2
print(random_list)
```

```python
random_list_log = np.log(random_list)
```

```python
print(random_list_log)
```

```python
# boolean (true/false) index our array to print out all even numbers
even_list = random_list[random_list % 2 == 0]
print(even_list)
```
